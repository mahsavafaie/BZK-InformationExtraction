## Analyze dataset

```python
import pandas as pd

df = pd.read_excel('transcription_BZK.xlsx')
df
```

```python
# remove Unnamed columns
df = df.loc[:,~df.columns.str.startswith('Unnamed:')]
df
```

```python
# analyze for each column how many non NaN values are present
print(df.notnull().sum() / len(df) * 100)
```

```python
# add '.jpg' to the file name
df.loc[:, 'Filename'] = df.loc[:, 'Filename'].apply(lambda x: x + '.jpg')

# rename filename to file_name
df = df.rename(columns={'Filename': 'file_name'})

df
```

```python
# check that all files exists on disk
import os
files_which_do_not_exist = []
for i in df['file_name']:
    file_name = './IE-GT/' + str(i)
    if not os.path.exists(file_name):
        files_which_do_not_exist.append(file_name)
print(len(files_which_do_not_exist))
print(files_which_do_not_exist)
```

```python
# show disjunct values for each column which has less than 50 unique values
for column in df.columns:
    if len(df[column].unique()) < 50:
        print(column)
        print(df[column].unique())
```

```python
#selected_columns = ['Filename', 'CompensatioOffice1', 'extra remarks 1', 'BZKNr', 
#                    'ApplicantFirstName', 'ApplicantLastName', 'ApplicantBirthName', 'ApplicantBirthDate', 'ApplicantBirthPlace', 'ApplicantCurrentAddress', 'ApplicantDeathDate', 'ApplicantDeathPlace',
#                    'VictimFirstName',    'VictimLastName',    'VictimBirthName',    'VictimBirthDate', 'VictimBirthPlace',       'VictimDeathDate', 'VictimDeathPlace']
selected_columns = ['file_name', 'CompensationOffice1', 'BZKNr', 
                    'ApplicantFirstName', 'ApplicantLastName', 'ApplicantBirthName', 'ApplicantBirthDate', 'ApplicantBirthPlace', 'ApplicantCurrentAddress',
                    'VictimFirstName',    'VictimLastName',    'VictimBirthName',    'VictimBirthDate', 'VictimBirthPlace',       'VictimDeathDate', 'VictimDeathPlace']
df_selected = df[selected_columns]
df_selected
```

```python
# replace NaN values with empty string
df_selected = df_selected.fillna('')

# remove all whitespace from cells
df_selected = df_selected.map(lambda x: x.strip() if isinstance(x, str) else x)
df_selected

df_selected
```

## Create Dataset

- First split it into train validation test

```python
# split in training , validation and test set (80%, 10%, 10%) 
# stratify by 'Layout class'
from sklearn.model_selection import train_test_split
train, test = train_test_split(df_selected, test_size=0.20, random_state=42, stratify=df_selected['Layout class'])
valid, test = train_test_split(test, test_size=0.50, random_state=42, stratify=df_selected['Layout class'])

train.shape, valid.shape, test.shape
```

```python
# create a ImageFolder (huggingface) image dataset
import shutil
import os
import csv

def create_image_folder_dataset(df, folder_name):
    if not os.path.exists(folder_name):
        os.makedirs(folder_name)
    for index, row in df.iterrows():
        file_name = './IE-GT/' + str(row['file_name'])
        if os.path.exists(file_name):
            shutil.copy(file_name, folder_name)
        else:
            print(f"File {file_name} does not exist")
    df.to_csv(folder_name + '/metadata.csv', index=False, quoting=csv.QUOTE_NONNUMERIC)

create_image_folder_dataset(train, './inferable/data/BZK/train')
create_image_folder_dataset(valid, './inferable/data/BZK/valid')
create_image_folder_dataset(test, './inferable/data/BZK/test')
```
